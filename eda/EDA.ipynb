```python
import pandas as pd

df = pd.read_csv("../data/movies_raw.csv")

print(df.shape)
df.head(10)
```

```python
print("Missing values per column:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```

```python
df['Awards']= pd.to_numeric(df['Awards'], errors = 'coerce')
df['Nominations'] = pd.to_numeric(df['Nominations'], errors='coerce')

df = df.dropna(subset=['Awards', 'Nominations'])
df['Awards'] = df['Awards'].astype(int)
df['Nominations'] = df['Nominations'].astype(int)

print(df.shape)
print(df.dtypes)
```

```python
df.describe().round(2)
```

```python
df['Decade'] = (df['Year'] // 10) * 10

decade_counts = df['Decade'].value_counts().sort_index()
print(decade_counts)
```

```python
top_awarded = df.sort_values('Awards', ascending=False).head(10)
top_awarded[['Film', 'Year', 'Awards', 'Nominations']]
```

```python
df['Win_Rate'] = df['Awards'] / df['Nominations']

print(df['Win_Rate'].describe())
```

```python
correlation = df['Nominations'].corr(df['Awards'])
print(f"Correlation between Nominations and Awards: {correlation:.2f}")
```

```python
decade_stats = df.groupby('Decade')[['Awards', 'Nominations']].mean().round(2)
print(decade_stats)
```

```python
df.to_csv("../data/movies_clean.csv", index=False)
print(f"Saved {len(df)} cleaned movies to movies_clean.csv")
```

## EDA Findings

- Dataset: 1,279 Academy Award-winning films (1934–2025), scraped from Wikipedia
- Award-winning films peaked in the 1940s (166 films) and are least common in the 1930s and 2020s (data still incomplete for 2020s)
- Average win rate (awards won / nominations received) is 64%, with a median of 50%
- Nominations have increased over time (3.44 avg in 1930s → 4.41 avg in 2020s), while awards per film stayed relatively flat (~1.6-1.7)
- Nominations and Awards are moderately correlated (r = 0.69) — more nominations generally means more wins, but it's not a guarantee
- Most-awarded films: Ben-Hur, Titanic, and The Lord of the Rings: The Return of the King (11 awards each)
